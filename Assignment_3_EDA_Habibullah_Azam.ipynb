```python
import pandas as pd

df = pd.read_csv("student-mat.csv", sep=";")

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.histplot(data=df, x="G3", bins=10, kde=True)

plt.title("Distribution of Final Grades (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()
```

### Interpretation of Final Grades

The histogram shows the distribution of students' final grades (G3). The average grade is 10.42 out of 20, while the median is 11. Grades range from 0 to 20. The graph helps us understand which grade ranges are more common among students.

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x="age", bins=8, kde=True)

plt.title("Distribution of Student Ages")
plt.xlabel("Age")
plt.ylabel("Number of Students")

plt.show()
```

### Interpretation of Student Ages

The histogram shows the age distribution of students in the dataset. It helps us identify the most common age group and understand the variation in student ages. This information helps describe the population included in the study.

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="sex")

plt.title("Number of Students by Gender")
plt.xlabel("Gender (F = Female, M = Male)")
plt.ylabel("Number of Students")

plt.show()
```

### Interpretation of Gender Distribution

The bar chart compares the number of male and female students in the dataset. It helps us understand the gender composition of the students and whether one group is larger than the other.

```python
plt.figure(figsize=(8,4))

sns.boxplot(x=df["G3"])

plt.title("Boxplot of Final Grades")
plt.xlabel("Final Grade (G3)")

plt.show()
```

### Interpretation of Final Grades Boxplot

The boxplot shows the spread and median of students' final grades. The median grade is 11, and the middle 50% of grades lie between 8 and 14. Grades far from the main group may be possible outliers and should be investigated further.

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="G3")

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x="studytime", y="G3")

plt.title("Study Time vs Final Grades")
plt.xlabel("Study Time (1 = Lowest, 4 = Highest)")
plt.ylabel("Final Grade (G3)")

plt.show()
```

### Interpretation of Study Time vs Final Grades

The boxplot compares students' final grades across different study-time groups. It helps us examine whether students who spend more time studying tend to achieve higher grades. However, study time alone does not determine academic performance.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x="absences", y="G3")

plt.title("Absences vs Final Grades")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")

plt.show()
```

### Interpretation of Absences vs Final Grades

The scatter plot shows the relationship between the number of student absences and final grades. Each point represents a student. The graph helps us identify whether students with more absences tend to have lower grades and whether any unusual observations are present.

```python
plt.figure(figsize=(12,8))

corr = df.corr(numeric_only=True)

sns.heatmap(corr, cmap="coolwarm", annot=True, fmt=".2f")

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

### Interpretation of Correlation Heatmap

The correlation heatmap shows the strength and direction of relationships between numerical variables. Positive values indicate that two variables tend to increase together, while negative values indicate an opposite relationship. Values close to zero suggest a weak linear relationship. The correlations between G1, G2, and G3 are particularly useful for understanding student performance.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="studytime",
    y="G3",
    hue="sex"
)

plt.title("Study Time, Final Grades and Gender")
plt.xlabel("Study Time (1 to 4)")
plt.ylabel("Final Grade (G3)")

plt.show()
```

### Interpretation of Multivariate Analysis

This scatter plot compares study time and final grades for male and female students. Different colors represent gender groups. It helps us examine whether the relationship between study time and final grades appears similar across the groups. The plot shows association, not proof of causation.

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["absences"])
plt.title("Boxplot of Student Absences")
plt.xlabel("Number of Absences")
plt.show()
```

### Interpretation of Student Absences

The boxplot shows the distribution of student absences and highlights possible outliers. Some students may have a much higher number of absences than others. These values should be investigated to understand whether they represent genuine student behavior or possible data issues. Unusual values should not be removed without a valid reason.
